### Q) Multiple Questions to ask
#### Find all the unique windspeed values in the data
#### The number of time the weather is exactly "Clear"
#### The number of time the WindSpeed was exactly 4 km/h
#### Find out all the null values in the data

```python
### Find all the unique windspeed values in the data
print("All the unique windspeed values in the data: ",data['Wind Speed_km/h'].unique())
print("\n")
### The number of time the weather is exactly "Clear"
print("The number of time the weather is exactly 'Clear': ", data.loc[data.Weather=='Clear','Weather'].count(), " times")
print("\n")
### The number of time the WindSpeed was exactly 4 km/h
print("The number of time the WindSpeed was exactly 4 km/h ",data.loc[data["Wind Speed_km/h"]==4,'Wind Speed_km/h'].count() ," times")
print("\n")
### Find out all the null values in the data
display("find null/missing values: ",data.isna().sum())
print("\n")
```

```python
### Rename the column 'Weather' to  'Weather Condition'
data.rename(columns={'Weather':'Weather Condition'},inplace=True)
display("Nouvelles colonnes: ",data.columns)
print("\n")

### Mean of Visibility columns
print("Visibility Mean: ",data["Visibility_km"].mean())
print("\n")
### Standard deviation of Presure column
print("Standard deviation of Pressure: ",data["Press_kPa"].std())
print("\n")
### Variance of relative humidity
print("Relative humidity variance: ",data["Rel Hum_%"].var())
print("\n")
### find all instances where snow was recored
display("all instances of snow:",data[data["Weather Condition"].str.contains('Snow')])
print("\n")
```

```python
### All instances where Wind Speed is above 24 and Visibility is 25
display("instances where Wind Speed is above 24 and Visibility is 25: ",data[(data["Wind Speed_km/h"]>24) & (data["Visibility_km"]==25)])
print("\n")
### Mean Value of each column against Weather Condition
display("Mean Value of each column against Weather Condition: ",data.groupby("Weather Condition").mean(numeric_only=True))
print("\n")
```

### Min and Max of each column against each Weather condition

```python

display("Min  of each column against Weather Condition: ",data.groupby("Weather Condition").min())
print("\n")
display("Max  of each column against Weather Condition: ",data.groupby("Weather Condition").max())
print("\n")
```

```python
plt.plot(data.loc[data["Weather Condition"]=='Fog',["Wind Speed_km/h","Weather Condition"]].groupby("Wind Speed_km/h").count(),color='r')
plt.xlabel(" occurence pr Wind speed Value")
plt.ylabel("Wind Speed Values")
```

### All instances when Weather is 'Clear' or Visibility > 40

```python
display(data[(data["Weather Condition"]=='Clear') & (data["Visibility_km"]>40)])
```

### All instances when Weather is 'Clear' and Relative Humidity greater than 50
### or
### Visibility > 40

```python
display(data[((data["Weather Condition"]=='Clear') & (data["Rel Hum_%"]>50)) | (data["Visibility_km"]>40)])
```
